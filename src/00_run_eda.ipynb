Do SNAP-eligible days induce higher demand?

```python
STATE = "TX"

snap_days_wi = calendar.loc[calendar[f"snap_{STATE}"] == 1,"d"].values
snap_wi_sales = unit_sales.loc[unit_sales.state_id ==STATE, [col for col in unit_sales.columns if (col in snap_days_wi)]]
nonsnap_wi_sales = unit_sales.loc[unit_sales.state_id ==STATE, [col for col in unit_sales.columns if ((col not in snap_days_wi) and col.startswith("d_"))]]
print(f"The average demand on SNAP-eligible days in {STATE} is {np.mean(snap_wi_sales.sum(axis=0)):.2f}.")
print(f"The average demand on non SNAP-eligible days in {STATE} is {np.mean(nonsnap_wi_sales.sum(axis=0)):.2f}.")
```

## Plot the demand-price curve to test the validity of linear assumption

For FOODS product category the negative linear relationship between price and quantity does not hold when we consider the total time horizon. Neither on department-level for a shorter time horizon: prices were found to be fixed on a 1Y time horizon for foods departments (sub-categories). The negative price-demand relationship does not hold for products with the longest history, rather the avg. weekly prices were stable and state-dependent.

**TL; DR: Choosing the 3 foods products that have the most price points over the history yields negative demand-price relationships in a given state.**

```python
unique_prices = (
    prices[prices["item_id"].str.startswith("FOODS")]
    .groupby(["store_id", "item_id"])["sell_price"]
    .nunique()
    .rename("n_unique_prices")
    .reset_index()
    .sort_values("n_unique_prices", ascending=False)
)

unique_prices.head(50)
```

```python
# get product-level demand for FOODS per day
# from EDA, the prices were founds to be state-dependent
foods_demand = unit_sales[unit_sales.state_id == "TX"]
foods_demand = foods_demand.loc[foods_demand.cat_id == 'FOODS',["item_id"] + [col for col in foods_demand.columns if col.startswith("d_")]]
q = foods_demand.groupby(by=["item_id"]).mean().reset_index()
```

```python
product_list = ["FOODS_1_096","FOODS_1_129"]
```

```python
q = q[q.item_id.isin(product_list)]
```

```python
merged_df = calendar.merge(prices, how="left", on="wm_yr_wk")
merged_df = merged_df[merged_df.item_id.isin(product_list)]
merged_df = merged_df.loc[merged_df.store_id.str.startswith("TX")]
```

```python
p = (
    merged_df
    .groupby(by=["item_id", "d"])["sell_price"]
    .mean()
    .reset_index()
    .assign(d_num=lambda df: df["d"].str.extract(r'(\d+)').astype(int))
    .sort_values(["item_id", "d_num"])
    .drop(columns="d_num")
)
```

```python
q_long = q.reset_index().melt(id_vars="item_id", var_name="d", value_name="quantity")
q_long.head()
```

```python
pq = p.merge(q_long,how="left", on=["d","item_id"])
pq = pq.dropna()
```

Optionally restrict the analysis to the last x weeks. In the final plot this was not done as in the AIDS paper of Deaton et al. (1980) time series data is used to estimate the demand equations.

```python
prices_demand = pq.drop(columns=["d"], inplace=False).groupby(by=["item_id","sell_price"]).mean().reset_index()
```

```python
sns.scatterplot(data=prices_demand, x="sell_price", y="quantity", hue="item_id")
plt.xlabel("Sell Price (USD)")
plt.ylabel("Avg. Sold Quantity")
# plt.title("Negative price-demand relationship holds for most active products")
plt.savefig(FIGURES_PATH + "pq_pattern.pdf", format="pdf", bbox_inches="tight")
```
